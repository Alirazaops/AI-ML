Raw_Data_To_Clean_Data

```python
import pandas as pd
import numpy as np
import os

emp = pd.read_excel(r"/Users/ali/Downloads/Rawdata.xlsx")
# Clean Name and Domain columns
for col in ['Name', 'Domain', 'Salary']:
    emp[col] = emp[col].str.replace(r'\W','',regex=True)

# Extract numbers from Age and Exp columns
for col in ['Age', 'Exp']:
    emp[col] = emp[col].str.extract('(\d+)')

# Fill missing values
for col in ['Age', 'Exp']:
    emp[col] = emp[col].fillna(np.mean(pd.to_numeric(emp[col])))

emp['Location'] = emp['Location'].fillna(emp['Location'].mode()[0])

print(emp)
print('###############################################################')

for col in ['Age', 'Exp', 'Salary']:
     emp[col] = emp[col].astype(int)

for col in ['Name','Domain','Location']:
     emp[col] = emp[col].astype('category')

emp.info()

print('###############################################################')

emp.to_csv('Clean_Data.csv')

os.getcwd()


```
